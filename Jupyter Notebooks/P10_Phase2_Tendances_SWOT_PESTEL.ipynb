## 1. Tendances actuelles du marché et des consommateurs

### 1.1 Un marché mondial mature qui continue de croître, mais dont la croissance ralentit

- Le marché mondial du jeu vidéo est estimé entre **189 et 205 milliards $** en 2026 selon Newzoo (fourchette large selon périmètre matériel/logiciel inclus ou non), pour une base de **3,6 à 3,7 milliards de joueurs**. *(Source : Newzoo Global Games Market Report 2026)*
- Signal important : la **croissance du nombre de joueurs (+4,4%) dépasse celle des revenus (+3,4%)** → la dépense moyenne par joueur payant recule légèrement (~120 $/an). Le marché gagne en volume mais pas proportionnellement en valeur. *(Source : Newzoo/SQ Magazine, 2026)*
- Le modèle économique repose de plus en plus sur la **rétention plutôt que l'acquisition** : achats intégrés (in-game purchases) = 74,4 Md$ en 2025 (+4,6%), abonnements type Game Pass = 11-13 Md$ pour ~120 millions d'abonnés. Une part croissante du chiffre d'affaires vient d'un petit nombre de jeux « live service » déjà installés (World of Warcraft, Diablo IV, Overwatch, Roblox). *(Source : Newzoo, Sensor Tower, 2026)*

### 1.2 Le marché français : solide, mais un joueur plus âgé qu'ailleurs

*(Source : SELL / Médiamétrie, bilan 2025, publié avril 2026)*

| Indicateur | Valeur 2025 | Évolution vs 2024 |
|---|---|---|
| Chiffre d'affaires total | 5,856 Md€ | +2,9% |
| Nombre de joueurs (1x/an) | 40 millions | — |
| Joueurs hebdomadaires | 76% des joueurs | +1 pt |
| Âge moyen d'un joueur | 40 ans | +4 pts (vieillissement) |
| Temps de jeu hebdomadaire moyen | 6h25 | — |
| Écosystème **console** | 2,552 Md€ (44% du marché) | stable |
| Écosystème **mobile** | 1,792 Md€ (30% du marché) | +11% |
| Écosystème **PC gaming** | 1,512 Md€ (26% du marché) | stable |

→ Pour un projet AAA (généralement pensé console/PC), le marché français reste porté à 70% par ces deux écosystèmes, mais c'est le **mobile qui capte toute la dynamique de croissance** cette année.

### 1.3 Préférences de genre : nos données historiques restent d'actualité

Sur le plan mondial, le classement des genres les plus pratiqués en 2025-2026 (part des joueurs actifs mensuels) est : **Shooter (51,8%) > Action-Aventure (48,8%) > Sport (36,3%)**. *(Source : Statista via Blog du Modérateur, oct. 2025)*

C'est cohérent avec ce que montrent nos propres données historiques (PJ2, ventes cumulées 1980-2016+) : Action, Sports et Shooter sont déjà les 3 premiers genres en volume de ventes (cf. Phase 1). Le marché n'a donc pas connu de rupture de préférence de genre depuis 15 ans — un signal de stabilité rassurant pour un choix de genre côté UOI Games, mais aussi un indice que ces catégories sont les plus concurrentielles.

### 1.4 Le joueur type n'est plus qui l'on croit

- Âge moyen du joueur : **37 ans** aux États-Unis (ESA 2026), **40 ans** en France (SELL 2025) — le gaming s'est nettement « vieilli » et n'est plus une pratique d'adolescents.
- Quasi-parité hommes/femmes : selon les sources, entre **45% et 52% de femmes** parmi les joueurs dans le monde et en France/Europe — la fourchette varie selon la méthodologie, mais toutes convergent vers une quasi-parité, loin du stéréotype du joueur homme jeune. *(Sources : ESA 2025-2026, Newzoo 2022, enquêtes UK/DE convergentes)*
- La pratique est devenue **cross-générationnelle** : aux États-Unis, près d'1 senior (80 ans et +) sur 3 joue chaque semaine, et 81% des parents joueurs jouent avec leurs enfants. *(Source : ESA Essential Facts 2026)*

**Implication pour UOI Games :** le ciblage « standard » (homme, 18-34 ans) n'est plus le seul segment pertinent ni forcément le plus différenciant — voir Phase 3 pour l'analyse chiffrée du/des segment(s).


## 2. Tendances futures et émergentes

### 2.1 Convergence des plateformes et cloud gaming
Selon le BCG Video Gaming Report 2026, la frontière console / PC / mobile s'efface au profit d'un modèle **« hardware-agnostic »** porté par le cloud : 60% des joueurs ont déjà testé le cloud gaming, 80% d'entre eux avec une expérience positive. L'ouverture programmée des stores mobiles (baisse des commissions) va aussi changer les modèles de distribution. *(Source : BCG, 2026)*

**Implication :** une stratégie mono-plateforme devient plus risquée à moyen terme — c'est un point que l'évaluateur pourra challenger en soutenance (cf. question anticipée sur la restriction de segment/plateforme).

### 2.2 L'IA générative transforme (lentement) l'économie de la production
- Les outils IA (génération de textures, concept art, animation via Move.AI/DeepMotion, doublage multilingue simultané par IA) réduisent déjà les coûts de certaines étapes de production de 30 à 40% dans les studios qui les ont intégrés. *(Source : Innovecs Games, 2026)*
- Une estimation (Morgan Stanley, relayée 2026) évoque un potentiel de réduction des coûts de développement proche de -50% à terme, mais la réalité observée aujourd'hui reste incrémentale : l'IA change surtout la répartition du temps des équipes créatives, pas encore le budget total.
- **Corollaire :** dans un marché saturé de contenu généré par IA, la « patte humaine » (finition artistique, narration) devient un argument de différenciation et un poste de coût qui ne baisse pas.

### 2.3 L'explosion des budgets AAA — un risque structurant pour le projet
- Les budgets de développement AAA (hors marketing) se situent aujourd'hui couramment entre **80 M$ et 300 M$+**, certains projets (Call of Duty, Red Dead Redemption 2) ayant dépassé les 300-700 M$ en comptant marketing et post-lancement. *(Sources : Bloomberg/Jason Schreier 2026, UK CMA — dossier Microsoft/Activision, Omdia)*
- Cette hausse est portée avant tout par les salaires et la durée des cycles de développement (3 à 6 ans, 200 à 1000+ personnes), plus que par la seule inflation technologique.

**Implication directe pour la recommandation finale (Phase 5) :** un premier titre AAA chez UOI Games devra vraisemblablement se positionner sur un **budget « AAA d'entrée de gamme »** (plutôt bas de la fourchette 80-150 M$) avec un genre déjà maîtrisé en interne, plutôt que de viser d'emblée un standard à 300 M$+ réservé aux studios avec un historique de blockbusters.

### 2.4 Un catalyseur conjoncturel à noter : GTA VI
Newzoo anticipe que la sortie de GTA VI (prévue fin 2026) sera le principal moteur de croissance du segment console/aventure cette année, au point que le marché console reculerait en son absence. C'est un signal fort de **concentration de la valeur sur un très petit nombre de titres évènementiels** — un contexte concurrentiel à ne pas sous-estimer pour un nouvel entrant AAA. *(Source : Newzoo Global Games Market Report 2026, relayé par Adgully, sept. 2026)*

## 4. Analyse PESTEL

| Facteur | Constats clés | Impact pour UOI Games |
|---|---|---|
| **Politique** | Multiplication des cadres réglementaires nationaux divergents sur le jeu vidéo (UE, UK, Brésil, États américains) ; tensions géopolitiques et commerciales pouvant affecter les chaînes d'approvisionnement composants | Anticiper une conformité multi-juridictions dès la conception si diffusion internationale |
| **Économique** | Pouvoir d'achat des ménages sous tension en France (contexte "défavorable" cité par le SELL) malgré un marché résilient ; hausse des coûts de composants mémoire PC ; budgets AAA en forte inflation (salaires, durée dev) | Nécessité d'un modèle économique robuste à un budget maîtrisé plutôt qu'un pari "tout ou rien" à 300 M$+ |
| **Socioculturel** | Joueur moyen vieilli (37-40 ans), quasi-parité hommes/femmes, pratique cross-générationnelle (parents/enfants, seniors) ; recherche de connexion sociale (78% des joueurs associent jeu vidéo et lien social) | Un ciblage marketing "ado masculin" serait désaligné avec la réalité du marché ; opportunité de segments sous-exploités (voir Phase 3) |
| **Technologique** | IA générative (assets, animation, doublage multilingue), essor du cloud gaming (60% l'ont testé), convergence console/PC/mobile, moteurs cross-plateforme | Opportunité de limiter les coûts de portage multi-plateforme et de production via l'IA, tout en gardant une direction artistique différenciante |
| **Écologique** | Enjeux croissants (mais encore peu réglementés spécifiquement jeu vidéo) autour de l'empreinte carbone du cloud gaming et des data centers IA ; pression RSE croissante des grands éditeurs | Facteur secondaire à ce stade pour un premier titre, mais à surveiller si le studio grandit |
| **Légal** | Digital Fairness Act (UE, adoption attendue fin 2026/2027) visant une interdiction des loot boxes accessibles aux mineurs ; PEGI 16 quasi-systématique dès présence d'achats aléatoires depuis juin 2026 ; Belgique/Pays-Bas déjà interdits, Brésil interdit dès mars 2026 ; procédures ouvertes contre des éditeurs (ex. AGCM Italie vs Activision Blizzard, janv. 2026) | Si le modèle économique envisagé inclut des mécaniques de type loot box, il doit être conçu "compliant by design" (prix réels affichés, pas de monétisation ciblant les mineurs) pour éviter un risque juridique et réputationnel dès le lancement |

*(Sources : SELL 2025/2026, thestar.com.my juil. 2026, blog.promise.legal juil. 2026, shattered.io juil. 2026, esportslegal.news déc. 2025, BCG 2026)*


## 5. Synthèse & transition vers la Phase 3

**À retenir pour la suite du projet :**
1. Le marché ne récompense plus la seule acquisition de nouveaux joueurs mais la **rétention** → penser dès la conception à une boucle de contenu/replay, pas seulement à la sortie du jeu.
2. Le joueur "moyen" a changé (plus âgé, mixte, cross-générationnel) → la Phase 3 doit tester des hypothèses de segment **au-delà** du stéréotype 18-34 ans masculin.
3. Les genres historiquement dominants (Action, Sport, Shooter) sont validés à la fois par nos données internes (PJ1/PJ2) et par les tendances 2025-2026 — mais ce sont aussi les plus saturés : le choix de segment devra arbitrer entre **taille de marché prouvée** et **différenciation concurrentielle**.
4. Toute mécanique de monétisation de type loot box doit être conçue en anticipant la régulation européenne à venir.
5. Le budget cible doit rester réaliste au regard des standards AAA actuels (80-300 M$+) compte tenu de l'absence de track record AAA d'UOI Games.
